<p style="color:#333333; font-size:15px; line-height:1.8; text-align:justify; overflow-wrap:anywhere; margin:0 0 20px 0;">A key objective of this programme is to overcome syntax-recall difficulties and develop the ability to independently solve coding problems under interview conditions. <h3 style="color:#1f4e79; margin:0 0 12px 0; font-size:20px;">1. Four Stages of Coding Proficiency</h3>
<div style="width:100%; max-width:100%; min-width:0; box-sizing:border-box; background-color:#f8fbff; border:1px solid #bfdbfe; border-radius:6px; padding:14px; margin:0 0 10px 0;">
<h4 style="color:#1f4e79; margin:0 0 6px 0; font-size:17px;">Stage 1 — Recognition</h4>
<p style="color:#333333; font-size:15px; line-height:1.8; text-align:justify; overflow-wrap:anywhere; margin:0;">Recognise Python syntax, understand code snippets and identify the purpose of commonly used functions and constructs.</p>
</div>
<div style="width:100%; max-width:100%; min-width:0; box-sizing:border-box; background-color:#ffffff; border:1px solid #bfdbfe; border-radius:6px; padding:14px; margin:0 0 10px 0;">
<h4 style="color:#1f4e79; margin:0 0 6px 0; font-size:17px;">Stage 2 — Syntax Recall</h4>
<p style="color:#333333; font-size:15px; line-height:1.8; text-align:justify; overflow-wrap:anywhere; margin:0;">Write commonly used syntax from memory, including loops, functions, collections, Pandas operations and exception handling.</p>
</div>
<div style="width:100%; max-width:100%; min-width:0; box-sizing:border-box; background-color:#f8fbff; border:1px solid #bfdbfe; border-radius:6px; padding:14px; margin:0 0 10px 0;">
<h4 style="color:#1f4e79; margin:0 0 6px 0; font-size:17px;">Stage 3 — Independent Implementation</h4>
<p style="color:#333333; font-size:15px; line-height:1.8; text-align:justify; overflow-wrap:anywhere; margin:0;">Translate problem statements into algorithms and implement complete solutions independently, including debugging and edge-case handling.</p>
</div>
<div style="width:100%; max-width:100%; min-width:0; box-sizing:border-box; background-color:#ffffff; border:1px solid #bfdbfe; border-radius:6px; padding:14px; margin:0 0 20px 0;">
<h4 style="color:#1f4e79; margin:0 0 6px 0; font-size:17px;">Stage 4 — Interview Execution</h4>
<p style="color:#333333; font-size:15px; line-height:1.8; text-align:justify; overflow-wrap:anywhere; margin:0;">Solve coding problems within a defined time limit, explain the approach, discuss complexity and handle follow-up questions without depending on Google or GPT.</p>
